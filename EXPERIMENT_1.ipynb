```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch

model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# Set pad_token to eos_token for consistent generation behavior
tokenizer.pad_token = tokenizer.eos_token

prompt = "Define Artificial Intelligence is transforming education.."
input_ids = tokenizer.encode(
prompt,
return_tensors="pt",
add_special_tokens=True
)
attention_mask = torch.ones(
input_ids.shape,
dtype=torch.long
)

output = model.generate(
input_ids,
attention_mask=attention_mask,
max_length=100,
num_return_sequences=1,
no_repeat_ngram_size=2,
top_k=50,
top_p=0.95,
temperature=0.7,
pad_token_id=tokenizer.eos_token_id # Explicitly set pad_token_id for generation
)

generated_text = tokenizer.decode(
output[0],
skip_special_tokens=True
)
print(generated_text)
```

```python
user_prompt = input("Enter your prompt for text generation: ")

input_ids_user = tokenizer.encode(
    user_prompt,
    return_tensors="pt",
    add_special_tokens=True
)
attention_mask_user = torch.ones(
    input_ids_user.shape,
    dtype=torch.long
)

output_user = model.generate(
    input_ids_user,
    attention_mask=attention_mask_user,
    max_length=100,
    num_return_sequences=1,
    no_repeat_ngram_size=2,
    top_k=50,
    top_p=0.95,
    temperature=0.7,
    pad_token_id=tokenizer.eos_token_id
)

generated_text_user = tokenizer.decode(
    output_user[0],
    skip_special_tokens=True
)
print("\nGenerated text based on your prompt:")
print(generated_text_user)
```
