```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# ============================================================
# 1. Load CodeGen model
# ============================================================

model_name = "Salesforce/codegen-350M-mono"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(model_name)

# Use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("Model loaded successfully!")
print("Using device:", device)


# ============================================================
# 2. Code Generation Function
# ============================================================

def generate_code(prompt, max_new_tokens=80):

    # Convert prompt to tokens
    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    # Move input to GPU/CPU
    input_ids = inputs.input_ids.to(device)

    # Generate code
    output = model.generate(
        input_ids,
        max_new_tokens=max_new_tokens,
        pad_token_id=tokenizer.eos_token_id,
        do_sample=False
    )

    # Convert generated tokens back to text
    generated_text = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return generated_text


# ============================================================
# 3. Code Generation from Natural Language
# ============================================================

prompt1 = """
# Write a Python function to check if a number is prime

def is_prime(n):
"""

generated_function = generate_code(
    prompt1,
    max_new_tokens=80
)

print("Generated Function:")
print(generated_function)


# ============================================================
# 4. Debugging a Faulty Code
# ============================================================

buggy_code = """
# The following function should return the factorial of n,
# but has a bug. Fix it.

def factorial(n):
    result = 0
    for i in range(1, n + 1):
        result = result * i
    return result

# Corrected function:

def factorial_fixed(n):
"""

debug_suggestion = generate_code(
    buggy_code,
    max_new_tokens=60
)

print("\nDebug Suggestion:")
print(debug_suggestion)
```
